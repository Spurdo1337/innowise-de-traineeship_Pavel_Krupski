**1.** Given a random array, change the sign of elements whose values ​​are between 3 and 8

**3.** Construct the direct product of arrays (all combinations with each element). A two-dimensional array is provided as input.

```python
# "Direct product"/"декартово произведение" - комбинация всех со всеми, не арифметика
def cartesian_product_v2(arrays):
    grids = np.meshgrid(*arrays, indexing='ij')
    return np.stack(grids, axis=-1).reshape(-1, len(arrays))

result = cartesian_product_v2([[1, 2, 3], [4, 5]])
print(result)
```

**4.** Given two arrays, A (8x3) and B (2x2), find rows in A that contain elements from each row in B, regardless of the order of the elements in B.

```python
A = np.random.randint(0, 5, (8, 3))
B = np.random.randint(0, 5, (2, 2))

print("A:\n", A)
print("B:\n", B)

C = (A[..., np.newaxis, np.newaxis] == B)
rows = np.where(C.any((3, 1)).all(1))[0]

print("Индексы подходящих строк:", rows)
```

**5.** Given a 10x3 matrix, find rows of unequal values ​​(for example, the row [2,2,3] remains, the row [3,3,3] is deleted)

```python
Z = np.random.randint(0, 2, (10, 3))
print("До:\n", Z)

# Сравниваем каждый элемент строки с первым элементом этой же строки
E = np.all(Z == Z[:, 0:1], axis=1)

# Оставляем только те строки, где НЕ все элементы равны (инвертируем маску)
U = Z[~E]
print("После:\n", U)
```

* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
# диагональ: X[0,0]=1, X[1,1]=0, X[2,2]=3
# произведение ненулевых: 1 * 3 = 3

def diag_product_pure(X):
    n = min(len(X), len(X[0]))
    result = 1 #чтобы избежать умножения на 0
    for i in range(n):
        val = X[i][i]
        if val != 0:
            result *= val
    return result

print(diag_product_pure(X))


X_np = np.array(X)
def diag_product_np(X):
    d = np.diagonal(X)
    return d[d != 0].prod()

print(diag_product_np(X_np))  
```

* __Task 2__: Two vectors x and y are given. Check if they define the same multiset.  
  For example, for x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) answer is True.

```python
#Вектор одномерный массив
#множество, в котором элементы могут повторяться

x = [1, 2, 2, 4]
y = [4, 2, 1, 2]
x_np = np.array(x)
y_np = np.array(y)

def same_multiset_pure(x, y):
    if len(x) != len(y):
        return False
    
    counts_x = {}
    for val in x:
        counts_x[val] = counts_x.get(val, 0) + 1
    
    counts_y = {}
    for val in y:
        counts_y[val] = counts_y.get(val, 0) + 1
    
    return counts_x == counts_y

print(same_multiset_pure(x, y))


def same_multiset_np(x, y):
    return np.array_equal(np.sort(x), np.sort(y))

print(same_multiset_np(x_np, y_np))
```

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero. 
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
x = [6, 2, 0, 3, 0, 0, 5, 7, 0]
x_np = np.array(x)

def max_after_zero_pure(x):
    candidates = []
    for i in range(1, len(x)):
        if x[i - 1] == 0:
            candidates.append(x[i])
    return max(candidates)

print(max_after_zero_pure(x))

def max_after_zero_np(x):
    mask = x[:-1] == 0
    return x[1:][mask].max()

print(max_after_zero_np(x_np))
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
x = [2, 2, 2, 3, 3, 3, 5]
x_np = np.array(x)

def rle_pure(x):
    values = []
    counts = []
    
    prev = x[0]
    count = 1
    
    for val in x[1:]:
        if val == prev:
            count += 1
        else:
            values.append(prev)
            counts.append(count)
            prev = val
            count = 1
    
    values.append(prev)
    counts.append(count)
    
    return values, counts

print(rle_pure(x))

def rle_np(x):
    change_points = np.where(x[1:] != x[:-1])[0] + 1
    idx = np.concatenate(([0], change_points, [len(x)]))
    
    values = x[idx[:-1]]
    counts = np.diff(idx)
    
    return values, counts

print(rle_np(x_np))
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
# матрица евклидовой дистанции ...
# евклидова дистанция - число, которое отражает 
# "насколько два объекта похожи/непохожи" с учётом всех их признаков одновременно 
# а не просто сравнение по одному признаку за раз.
# distance(a, b) = √((a₁-b₁)² + (a₂-b₂)²)

X = [[0, 0], [1, 1], [2, 2]]
Y = [[0, 0], [3, 4], [1, 0]]

X_np = np.array(X)
Y_np = np.array(Y)

def euclidean_py(X, Y):
    n = len(X)
    m = len(Y)
    result = [[0] * m for _ in range(n)]
    
    for i in range(n):
        for j in range(m):
            s = 0
            for k in range(len(X[i])):
                s += (X[i][k] - Y[j][k]) ** 2
            result[i][j] = s ** 0.5
    
    return result

result_py = euclidean_py(X, Y)
rounded = [[round(val, 2) for val in row] for row in result_py]
print(rounded)


def euclidean_np(X, Y):
    diff = X[:, np.newaxis, :] - Y[np.newaxis, :, :]
    return np.sqrt((diff ** 2).sum(axis=2))

result_np = euclidean_np(X_np, Y_np)
print(np.round(result_np, 2))
```
